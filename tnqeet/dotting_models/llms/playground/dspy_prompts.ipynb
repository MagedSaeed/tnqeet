```python
from tnqeet.dotting_models.llms. models import OpenRouterArabicDotter, DetailedArabicDotingSignature
from tnqeet import remove_dots
```

```python
dotter = OpenRouterArabicDotter(
    dspy_cache=False,
    model='openrouter/anthropic/claude-sonnet-4.5'
)
dotter
```

```python
text = "أريد أن أتعلم البرمجة بلغة بايثون"
dotless_text = remove_dots(text)
text,dotless_text
```

```python
dotter.restore_dots(dotless_text)
```

```python
dotter.lm.history[-1]
```

```python
dotter.lm.inspect_history()
```
